```python
# Create column total amount
df['Total_Amount']= df['Quantity']*df['UnitPrice']

#Create column day
df['Day_name']= df['InvoiceDate'].dt.day_name()
df['Day_Type'] = df['InvoiceDate'].dt.dayofweek.apply(lambda x: 'Weekend' if x >= 5 else 'Weekday')
df ['Hour'] = df['InvoiceDate'].dt.hour
```

```python
#Month Trend
df['Month_Year']= df['InvoiceDate'].dt.to_period('M')

def get_time_category (Hour):
  if 7<= Hour <12:
    return 'Morning'
  elif 13<= Hour <17:
    return 'Afternoon'
  else:
    return 'Night'
df['Time_Category'] = df['Hour'].apply(get_time_category)
```

```python
# Question 1: What is the trend of total sales (revenue) per month?

# 1. Agregasi dan Urutkan dari Total Amount Terbesar (Descending)
month_trend = (
    df.groupby('Month_Year')['Total_Amount']
    .sum()
    .reset_index()
    .sort_values(by='Total_Amount', ascending=False)
)

# 2. Tampilkan dengan Warna Gradasi (Semakin Tinggi = Warna Semakin Pekat)
month_trend.style.background_gradient(subset=['Total_Amount'], cmap='Blues').format({'Total_Amount': '{:,.2f}'})
```

```python
# Question 1: What is the trend of total sales (revenue) per month?
# Mengganti 'Total_Revenue' menjadi 'Total_Amount'
trend_bulanan = df.groupby(df['Month_Year'].astype(str))['Total_Amount'].sum().reset_index()

sns.set_theme(style='white')
plt.figure(figsize=(10, 5))

ax = sns.lineplot(data=trend_bulanan, x='Month_Year', y='Total_Amount', marker='o', color='#2B5C8F', linewidth=2.5)

for index, row in trend_bulanan.iterrows():
    ax.text(row['Month_Year'], row['Total_Amount'] + 300, f"{int(row['Total_Amount']):,}", ha='center', fontsize=9, fontweight='bold')

plt.title('Monthly Total Sales (Revenue) Trend', fontsize=14, fontweight='bold', pad=15, loc='center')
plt.xlabel('Month-Year', fontsize=12)
plt.ylabel('Total Revenue', fontsize=12)
plt.xticks(rotation=45)

# Menghilangkan gridline
plt.grid(False)

# Menambah batas sumbu Y agar teks angka paling atas tidak terpotong
plt.ylim(top=trend_bulanan['Total_Amount'].max() * 1.1)

sns.despine()
plt.tight_layout()
plt.show()
```

Based on the chart, monthly revenue peaked in November 2011 at 18,000 and hit its lowest point in February 2011.This sharp surge is highly indicative of year end holiday shopping behavior or successful promotional campaign. There was also a drastic drop in December 2011. The root cause of this decline must be thoroughly investigated to develop strategic solutions and prevent future revenue leakage

```python
# Grouping total unit berdasarkan Month_Year
unit_per_month = df.groupby(df['Month_Year'].astype(str))['Quantity'].sum().reset_index()

plt.figure(figsize=(12, 6))
ax = sns.barplot(data=unit_per_month, x='Month_Year', y='Quantity', color='#4C72B0')

# Menambahkan label nilai di atas setiap batang
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f'{int(height):,}',
            (p.get_x() + p.get_width() / 2., height),
            ha='center', va='bottom',
            xytext=(0, 4),
            textcoords='offset points',
            fontsize=9
        )

# Memberi ruang ekstra di atas sumbu Y agar angka tidak terpotong
plt.ylim(top=unit_per_month['Quantity'].max() * 1.12)

plt.xlabel('Month-Year')
plt.ylabel('Total Unit Terjual')
plt.title('Monthly Total Units Sold', fontsize=14, fontweight='bold', pad=15)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Question 2: During which part of the day (Time Category) do customers transact most frequently?
df['Time_Category'].unique()
```

```python
time_order = ['Afternoon','Morning','Night']
trend_time = df ['Time_Category'].value_counts().reindex(time_order).reset_index()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

# 1. Cari nilai maksimum
max_val = trend_time['count'].max()

# 2. Buat skema warna dinamis (hanya nilai tertinggi yang berwarna)
colors = ['#2B5C8F' if val == max_val else '#B0BEC5' for val in trend_time['count']]

# 3. Plotting
ax = sns.barplot(
    data=trend_time,
    x='Time_Category',
    y='count',
    palette=colors
)

# Menambahkan label angka di atas setiap bar
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f'{int(height):,}',
            (p.get_x() + p.get_width() / 2., height),
            ha='center', va='bottom',
            xytext=(0, 4),
            textcoords='offset points',
            fontsize=10,
            fontweight='bold' if height == max_val else 'normal'
        )

# Menambah ruang batas atas sumbu Y agar teks tidak menempel garis atas
plt.ylim(top=max_val * 1.12)

plt.title('Number of Transactions by Time of Day', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Time of Day', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

The graph indicates that the number of transactions is highest in the afternoon—surpassing 2,000 orders—compared to the morning and evening, suggesting a preference for shopping between 1:00 PM and 5:00 PM. To sustain order volumes, we should align marketing campaigns, promotional push notifications, and customer support availability with this afternoon surge to maximize engagement and ensure efficient conversion rates when purchasing intent is at its peak. Meanwhile, to increase orders during other times, solutions such as special discounts—like "low-price add-on" offers, "buy-one-get-one" deals, or similar promotions—can be implemented to achieve a more even distribution of orders.

```python
# Question 3: Which day and day type (weekday vs. weekend) yield the highest average revenue per transaction?

df['Day_name'].unique()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Agregasi data
trend_1 = df.pivot_table(index=['Day_name', 'Day_Type'], values='Total_Amount', aggfunc='mean').reset_index()
trend_1 = trend_1.sort_values(by='Total_Amount', ascending=False)

# 2. Tentukan nilai maksimum dan minimum
max_val = trend_1['Total_Amount'].max()
min_val = trend_1['Total_Amount'].min()

colors = []
for val in trend_1['Total_Amount']:
    if val == max_val:
        colors.append('#2B5C8F')  # Paling Besar
    elif val == min_val:
        colors.append('#D95F02')  # Paling Kecil
    else:
        colors.append('#B0BEC5')  # Sisanya

# 4. Plotting
plt.figure(figsize=(8, 5))
ax = sns.barplot(
    data=trend_1,
    x='Day_name',
    y='Total_Amount',
    hue='Day_name',
    palette=colors,
    legend=False
)

# Menambahkan label nilai di atas setiap bar
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f'{height:,.2f}',
            (p.get_x() + p.get_width() / 2., height),
            ha='center', va='bottom',
            xytext=(0, 4),
            textcoords='offset points',
            fontsize=9,
            fontweight='bold' if height in [max_val, min_val] else 'normal'
        )

# Menambah batas atas sumbu Y agar teks angka tidak terpotong
plt.ylim(top=max_val * 1.12)

plt.title('Average Revenue per Transaction by Day', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Day Name', fontsize=10)
plt.ylabel('Average Revenue', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Agregasi data
trend_1 = df.pivot_table(index=['Day_name', 'Day_Type'], values='Total_Amount', aggfunc='mean').reset_index()

# 2. Urutkan hari secara logis dari Monday sampai Sunday
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
trend_1['Day_name'] = pd.Categorical(trend_1['Day_name'], categories=day_order, ordered=True)
trend_1 = trend_1.sort_values('Day_name')

# 3. Tentukan nilai maksimum dan minimum
max_val = trend_1['Total_Amount'].max()
min_val = trend_1['Total_Amount'].min()

# 4. Plotting Line Chart
plt.figure(figsize=(9, 5))
ax = sns.lineplot(
    data=trend_1,
    x='Day_name',
    y='Total_Amount',
    marker='o',
    markersize=8,
    color='#2B5C8F',
    linewidth=2.5
)

# 5. Tambahkan penanda warna khusus pada titik tertinggi & terendah + label angka
for index, row in trend_1.iterrows():
    val = row['Total_Amount']
    day = row['Day_name']

    # Beri warna merah untuk nilai terendah, biru untuk tertinggi, abu-abu untuk sisanya
    if val == max_val:
        color_point = '#2B5C8F'
        font_weight = 'bold'
    elif val == min_val:
        color_point = '#D95F02'
        font_weight = 'bold'
    else:
        color_point = '#B0BEC5'
        font_weight = 'normal'

    # Gambar titik (marker) dengan warna spesifik
    plt.plot(day, val, marker='o', markersize=10, color=color_point)

    # Tambahkan teks angka di atas/bawah titik
    ax.annotate(
        f'{val:,.2f}',
        (day, val),
        ha='center', va='bottom',
        xytext=(0, 6),
        textcoords='offset points',
        fontsize=9,
        fontweight=font_weight
    )

# Menambah batas atas sumbu Y agar teks angka tidak terpotong
plt.ylim(top=max_val * 1.12)

plt.title('Average Revenue per Transaction by Day', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Day Name', fontsize=10)
plt.ylabel('Average Revenue', fontsize=12)
plt.grid(False)  # Hilangkan gridline
sns.despine()     # Merapikan bingkai grafik
plt.tight_layout()
plt.show()
```

The graph shows a downward trend over time, with the highest revenue recorded on Thursday, when the number of orders reached 30. The day-to-day decrease is not particularly drastic. In terms of sales, the lowest number was recorded on Sunday. There are two possible reasons why Sunday has the lowest revenue. Perhaps the people who visit the store are just there to window shop and look around.

```python
# Question 4: What does the comparison map of transaction volumes across day types within each time category look like?

cross_table =pd.crosstab(df['Time_Category'],df['Day_Type'])
melt_table= cross_table.reset_index().melt(id_vars='Time_Category', value_vars=['Weekday','Weekend'], value_name='Total_Orders')

sns.barplot(data=melt_table  , x='Time_Category', y='Total_Orders', hue='Day_Type', palette='Dark2')
plt.title('Order Distribution: Time Category vs Day Type', fontsize=12, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

This bar chart shows that transaction volume in the afternoon is significantly higher than during the morning and evening hours, both on weekdays and weekends, with the number of orders on weekday afternoons peaking well above 1,750. This clear surge indicates the presence of a high-volume operational period during which customer demand is concentrated. To ensure seamless operations and maintain high service standards, it is highly recommended to optimize workforce management by allocating additional staff during afternoon peak hours. Increasing frontline or fulfillment personnel during this window will enable the team to process surging orders more rapidly, reduce response times, and ultimately elevate customer satisfaction.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Menyiapkan kanvas khusus (1 kolom di atas untuk Scatter & KDE, Violin di bawah)
fig = plt.figure(figsize=(16, 18))
gs = fig.add_gridspec(2, 2, height_ratios=[1, 2])

ax_violin = fig.add_subplot(gs[1, :]) # Violin Plot mengambil full baris bawah

country_order = df.groupby('Country')['Quantity'].median().sort_values(ascending=False).index

sns.violinplot(
    data=df,
    x='Quantity',
    y='Country',
    hue='Country',
    legend=False,
    order=country_order,
    palette='Blues_r',
    cut=0,           # Membatasi violin hanya pada rentang data asli (tidak melebar berlebih)
    ax=ax_violin
)
ax_violin.set_title('Violin Plot (Quantity Distribution across All Countries)', fontsize=12, fontweight='bold')
ax_violin.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Question 5: Excluding the United Kingdom, which five countries had the highest total purchase quantities of goods?

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Agregasi data Top 5 Negara (Excluding UK)
top_country = df[df['Country'] != 'United Kingdom'].groupby('Country')['Quantity'].sum().nlargest(5).reset_index()

# 2. Plotting
plt.figure(figsize=(9, 5))
ax = sns.barplot(
    data=top_country,
    x='Quantity',
    y='Country',
    color='#2B5C8F'
)

# 3. Menambahkan label angka di ujung setiap bar
for p in ax.patches:
    width = p.get_width()
    if width > 0:
        ax.annotate(
            f'{width:,.0f}',
            (width, p.get_y() + p.get_height() / 2.),
            ha='left', va='center',
            xytext=(7, 0),
            textcoords='offset points',
            fontsize=10,
            fontweight='bold'
        )

# 4. Merapikan batas sumbu X agar label tidak terpotong
max_val = top_country['Quantity'].max()
plt.xlim(right=max_val * 1.15)

plt.title('Top 5 Countries by Quantity Sold (Excluding UK)', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Total Quantity Sold', fontsize=11)
plt.ylabel('Country', fontsize=11)
plt.grid(axis='x', linestyle='--', alpha=0.5)

sns.despine()
plt.tight_layout()
plt.show()
```

Netherlands emerges as the leading global market with a total quantity sold approaching 2,500 units—supported by its higher median order size shown in the boxplot. Germany follows closely in second place (approx. 1,750 units), with EIRE, France, and Sweden safely rounding out the top five international revenue generators.
Since the Netherlands and Germany demonstrate both high total volume order distributions, logistics and marketing teams should prioritize these two regions for localized fulfillment strategies, targeted ad spend, and strategic cross-border shipping partnerships to lower delivery friction and boost customer retention.

```python
# Total transaksi setiap invoice pada masing-masing negara
invoice_total = (
    df.groupby(['Country', 'InvoiceNo'])['Total_Amount']
      .sum()
      .reset_index()
)
invoice_total
```

```python
# Rata-rata transaksi per negara
avg_transaction = (
    invoice_total.groupby('Country')['Total_Amount']
                 .mean()
                 .sort_values(ascending=False)
                 .reset_index()
)
avg_transaction.head(10)
```

```python
import matplotlib.pyplot as plt

top10 = avg_transaction.head(10)

# 1. Tentukan nilai maksimum dan minimum
max_val = top10['Total_Amount'].max()
min_val = top10['Total_Amount'].min()

# 2. Buat skema warna kustom
colors = []
for val in top10['Total_Amount']:
    if val == max_val:
        colors.append('#2B5C8F')
    elif val == min_val:
        colors.append('#D95F02')
    else:
        colors.append('#B0BEC5')

# 3. Plotting
plt.figure(figsize=(13, 6))
bars = plt.bar(top10['Country'], top10['Total_Amount'], color=colors)

# Add numerical values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        x=bar.get_x() + bar.get_width() / 2.,
        y=height + 3,
        s=f'{height:,.2f}',
        ha='center', va='bottom',
        fontsize=9,
        fontweight='bold' if height in [max_val, min_val] else 'normal'
    )

# Add padding to top of Y-axis to prevent label clipping
plt.ylim(top=max_val * 1.12)

plt.title('Top 10 Countries by Average Transaction Value', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Country', fontsize=12)
plt.ylabel('Average Transaction Value', fontsize=12)
plt.tight_layout()
plt.show()
```
